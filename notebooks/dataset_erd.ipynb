```python
erd = ERD()

erd.add_table(dfs['olist_customers_dataset'], 'Customers')
erd.add_table(dfs['olist_orders_dataset'], 'Orders')
erd.add_table(dfs['olist_order_items_dataset'], 'OrderItems')
erd.add_table(dfs['olist_products_dataset'], 'Products')
erd.add_table(dfs['olist_sellers_dataset'], 'Sellers')
erd.add_table(dfs['olist_order_payments_dataset'], 'Payments')
erd.add_table(dfs['olist_order_reviews_dataset'], 'Reviews')
erd.add_table(dfs['olist_geolocation_dataset'], 'Geolocation')
erd.add_table(dfs['product_category_name_translation'], 'CategoryTranslation')
```

```python
erd.create_rel('Customers', 'Orders', on='customer_id', left_cardinality='1', right_cardinality='*')
erd.create_rel('Orders', 'OrderItems', on='order_id', left_cardinality='1', right_cardinality='*')
erd.create_rel('Products', 'OrderItems', on='product_id', left_cardinality='1', right_cardinality='*')
erd.create_rel('Sellers', 'OrderItems', on='seller_id', left_cardinality='1', right_cardinality='*')
erd.create_rel('Orders', 'Payments', on='order_id', left_cardinality='1', right_cardinality='*')
erd.create_rel('Orders', 'Reviews', on='order_id', left_cardinality='1', right_cardinality='*')
erd.create_rel('Products', 'CategoryTranslation', on='product_category_name', left_cardinality='*', right_cardinality='1')
```

```python
erd.write_to_file(PROJECT_ROOT/'notebooks'/'olist_erd.txt')
```
